# Data Cleaning

```python
# Select Only required columns 
req_columns = ["Age","Gender","Branch","GHQ Score","Total Score (DASSD-I)","Total Score (DASSD-II)","Total Score (DASSD-III)","BRS Score"]
```

```python
# dataframe with required columns 
sub_data = data[req_columns].copy()
```

```python
sub_data.head()
```

```python
sub_data.isna().sum()
```

```python
sub_data[sub_data["BRS Score"]==0]
```

```python
sub_data["Age"].value_counts()
```

```python
sub_data["Age"] = [18 if Age not in (17,18,19,20,21) else Age for Age in sub_data["Age"]]
```

```python
sub_data.rename(columns={
    "GHQ Score":"GHQ",
    "Total Score (DASSD-I)": "Depression",
    "Total Score (DASSD-II)": "Anxiety",
    "Total Score (DASSD-III)": "Stress",
    "BRS Score": "BRS"
},inplace=True)
```

```python
sub_data.to_csv("../data/cleaned/processing.csv",index=False)
```
