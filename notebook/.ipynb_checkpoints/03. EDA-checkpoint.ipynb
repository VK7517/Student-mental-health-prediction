```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = pd.read_csv("../data/cleaned/processing.csv")
```

```python
data.head()
```

```python
data["Age"] = [18 if Age not in (17,18,19,20,21) else Age for Age in data["Age"]]
```

```python
data["Age"].value_counts()
```

```python
# Age vs Depression 
y = data.groupby("Age")["Depression"].mean()
plt.plot(y)
plt.title("Age vs Depression ")
plt.xlabel("Age Group")
plt.ylabel("Avg Depression Level")
plt.tight_layout()
```

```python
# Age vs Anxiety 
y = data.groupby("Age")["Anxiety"].mean()
plt.plot(y)
plt.title("Age vs Anxiety ")
plt.xlabel("Age Group")
plt.ylabel("Avg Anxiety Level")
plt.tight_layout()
```

```python
# Age vs Stress 
y = data.groupby("Age")["Stress"].mean()
plt.plot(y)
plt.title("Age vs Stress ")
plt.xlabel("Age Group")
plt.ylabel("Avg Stress Level")
plt.tight_layout()
```

```python
data["Gender"].value_counts()
```

```python
# Gender vs Depression
age_vs_gender = data.groupby("Gender")["Depression"].mean()
age_vs_gender = age_vs_gender.drop("Other")
plt.bar(age_vs_gender.index,age_vs_gender.values)
plt.title("Gender vs Depression")
plt.xlabel("Gender")
plt.ylabel("Avg Depression Level")
plt.tight_layout()
```

```python
# Gender vs Anxiety
age_vs_gender = data.groupby("Gender")["Anxiety"].mean()
age_vs_gender = age_vs_gender.drop("Other")
plt.bar(age_vs_gender.index,age_vs_gender.values)
plt.title("Gender vs Anxiety")
plt.xlabel("Gender")
plt.ylabel("Avg Anxiety Level")
plt.tight_layout()
```

```python
# Gender vs Stress
age_vs_gender = data.groupby("Gender")["Stress"].mean()
age_vs_gender = age_vs_gender.drop("Other")
plt.bar(age_vs_gender.index,age_vs_gender.values)
plt.title("Gender vs Stress")
plt.xlabel("Gender")
plt.ylabel("Avg Stress Level")
plt.tight_layout()
```
