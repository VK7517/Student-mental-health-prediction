```python
# Age Distribution
plt.figure(figsize=(7,5))
plt.hist(data["Age"],bins=4)
plt.title("Age Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("../figures/Age_Distribution.png")
```

```python
# Age vs Depression 
y = data.groupby("Age")["Depression"].mean()
plt.plot(y,marker="o")
plt.title("Age vs Depression ")
plt.xlabel("Age Group")
plt.ylabel("Avg Depression Level")
plt.tight_layout()
# plt.savefig("../figures/Age_vs_Depression.png")
```

```python
# Age vs Stress 
y = data.groupby("Age")["Stress"].mean()
plt.plot(y)
plt.title("Age vs Stress ")
plt.xlabel("Age Group")
plt.ylabel("Avg Stress Level")
plt.tight_layout()
```

```python
# Age vs Resilience (BRS) 
y = data.groupby("Age")["BRS"].mean()
plt.plot(y)
plt.title("Age vs Resilience Scale ")
plt.xlabel("Age Group")
plt.ylabel("Avg Resilience Level")
plt.tight_layout()
# plt.savefig("../figures/Age_vs_Resilience.png")
```

```python
# Resilience Level Distribution

plt.figure(figsize=(12,5))
params = {
    "edgecolor":"Black",
    "linewidth":1.5,
    "width":0.6
}
plt.pie(x=data["BRS_Level"].value_counts().values, labels =data["BRS_Level"].value_counts().index , autopct="%1.1f%%" , wedgeprops=params,colors=["#4E79A7", "#F28E2B", "#E15759","#76B7B2"])
plt.title("Percentage of Students as per Resilience Levels")
plt.tight_layout()
plt.savefig("../figures/Resilience_Distribution.png")
```

```python
# Resilience vs Depression 
resilience_vs_Depression = data.groupby("BRS_Level")["Depression"].mean()
count = resilience_vs_Depression
plt.figure(figsize=(8,7))
plt.bar(count.index,count.values,width=0.4)
plt.title("Resilience Level vs Depression ")
plt.xlabel("Resilience Level")
plt.ylabel("Depression Score")
plt.tight_layout()
plt.savefig("../figures/Resilience_vs_Depressioin.png")
```

# Correlation Heatmap

```python

cols = ["GHQ","Anxiety","Stress","Depression","BRS"] 
cor = data[cols].corr()
plt.figure(figsize=(8,7))
sns.heatmap(cor,
           annot=True)
plt.tight_layout()
plt.savefig("../figures/Correlation_Heatmap.png")
```

# Cluster Analysis 

```python
count = data["Cluster"].value_counts().sort_index()
plt.figure(figsize=(7,6))
plt.bar(count.index,count.values)
plt.title("Cluster Distribution")
plt.xlabel("Cluster Names")
plt.ylabel("Count of Cluster")
plt.tight_layout()
plt.savefig("../figures/Cluster_Distribution.png")
```

# Cluster Wise Distribution

```python
# Gender vs Cluster 
gender_vs_cluster = data.groupby(["Cluster","Gender"]).size()
gender_vs_cluster.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 9),
    color=["teal","orange","red"]
)

plt.title("Gender Distribution Across Clusters")
plt.xlabel("Cluster")
plt.ylabel("Number of Students")
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
plt.savefig("../figures/Cluster_with_Gender.png")
```

```python
# Age vs Cluster 
age_vs_cluster = (data.groupby(["Cluster","Age"]).size().unstack(fill_value=0))
ax = age_vs_cluster.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 9),
    color=["teal","orange","red","pink","brown"]
)

ax.legend(title="Age")
plt.title("Age Distribution Across Clusters")
plt.xlabel("Cluster with Age")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

# Key Insights 

### 1. Most responses were received from students aged 18–20 years, so the dataset mainly represents early undergraduate students. 

### 2. Students aged 18–21 years showed higher levels of depression than other age groups, suggesting that early college years may be more stressful.

### 3. Most students had Normal or Mild Anxiety, but around 15% of students had Severe Anxiety, which indicates that a considerable number of students may need mental health support.

### 4. Most students had Normal Resilience, while around 14% had Low Resilience, meaning some students may have difficulty coping with challenges and stress. 

### 5. Students with Low Resilience generally showed higher depression levels, suggesting that stronger coping ability may help reduce depressive symptoms. 

### 6. The correlation analysis showed that Resilience and Depression are strongly related, indicating that resilience plays an important role in students' mental wellbeing.

### 7. Cluster F contains only a small number of students, but it is the most critical group because these students have both high psychological distress and low resilience. They may require immediate counseling and continuous support.
